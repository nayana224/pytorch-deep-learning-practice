# 07-1. ViT Patch Embedding

**핵심 질문:** 이미지를 patch token sequence로 바꾸는 과정은 실제 tensor에서 어떻게 보이는가?

32×32 CIFAR-10 이미지를 4×4 patch로 나누어 `[B,3,32,32] → [B,64,D]` 변환을 확인합니다.

```python
import torch
import torch.nn as nn

x = torch.randn(2, 3, 32, 32)
patch_embed = nn.Conv2d(3, 64, kernel_size=4, stride=4)
y = patch_embed(x)
print('feature grid:', y.shape)
tokens = y.flatten(2).transpose(1, 2)
print('tokens:', tokens.shape)
```
